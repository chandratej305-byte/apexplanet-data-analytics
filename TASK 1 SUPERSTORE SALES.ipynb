```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\chand\Desktop\apexplanet-data-analytics\data\Sample - Superstore.csv",
    encoding="latin1"
)

df.describe()
```

```python
df.columns = df.columns.str.strip()
```

```python
print(df.columns.tolist())
```

```python
print("Missing values per column:\n", df.isnull().sum())

df['Postal Code'] = df['Postal Code'].fillna(0)
df.dropna(subset=['Order ID', 'Sales', 'Profit'], inplace=True)
```

```python

duplicate_count = df.drop(columns=['Row ID']).duplicated().sum()
print("Duplicate order lines found:", duplicate_count)

df = df.drop_duplicates(subset=df.columns.difference(['Row ID']))
print("Shape after removing duplicates:", df.shape)
```

```python
def remove_outliers_iqr(dataframe, column):
    Q1 = dataframe[column].quantile(0.25)
    Q3 = dataframe[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return dataframe[(dataframe[column] >= lower_bound) & (dataframe[column] <= upper_bound)]

df_clean = remove_outliers_iqr(df, 'Sales')
print("Shape before IQR:", df.shape)
print("Shape after IQR on Sales:", df_clean.shape)
```

```python
df_clean.to_csv('cleaned_superstore.csv', index=False)
print("Cleaned dataset saved successfully!")
```

```python
df_clean.info()
df_clean[['Sales', 'Quantity', 'Discount', 'Profit']].describe()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(df_clean['Sales'], bins=30, kde=True, ax=axes[0], color='steelblue')
axes[0].set_title('Distribution of Sales (Post-IQR)')

sns.boxplot(x=df_clean['Profit'], ax=axes[1], color='salmon')
axes[1].set_title('Boxplot of Profit')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(data=df_clean, x='Category', y='Sales', estimator=sum, ax=axes[0], palette='Blues_d')
axes[0].set_title('Total Sales by Category')

sub_profit = df_clean.groupby('Sub-Category', observed=False)['Profit'].sum().sort_values()
sub_profit.plot(kind='barh', ax=axes[1], color='teal')
axes[1].set_title('Total Profit by Sub-Category')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
numeric_cols = ['Sales', 'Quantity', 'Discount', 'Profit']
sns.heatmap(df_clean[numeric_cols].corr(), annot=True, cmap='RdBu_r', center=0, fmt='.2f')
plt.title('Correlation Matrix of Superstore Metrics')
plt.show()
```

```python

df_clean['Order Date'] = pd.to_datetime(df_clean['Order Date'], format='mixed')

monthly_trend = df_clean.set_index('Order Date').resample('ME')[['Sales', 'Profit']].sum()

plt.figure(figsize=(12, 5))
plt.plot(monthly_trend.index, monthly_trend['Sales'], label='Monthly Sales ($)', color='navy', linewidth=2)
plt.plot(monthly_trend.index, monthly_trend['Profit'], label='Monthly Profit ($)', color='green', linewidth=2)
plt.title('Monthly Sales & Profit Trend (2014–2017)')
plt.xlabel('Order Date')
plt.ylabel('Amount ($)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Key Insights from Exploratory Data Analysis (EDA)

1. **Discounts Erode Profitability:** The correlation heatmap confirms a strong negative relationship between `Discount` and `Profit`. Orders with heavy discounts consistently drag down overall profitability and generate net losses.
2. **Sub-Category Profit Imbalance:** The Sub-Category bar chart shows that **Tables** and **Bookcases** generate substantial losses despite solid sales volume, whereas **Phones**, **Paper**, and **Accessories** deliver the highest net profits.
3. **Seasonal Year-End Surge:** The monthly trend line reveals a clear seasonal cycle, with sales dipping in Q1 (January–February) and peaking sharply every **November and December** during fourth-quarter holiday and corporate purchasing.
4. **Right-Skewed Order Distribution:** After removing extreme outliers via the IQR method, the Sales histogram shows that the vast majority of standard orders fall under **$200**, with order frequency dropping steadily as transaction size increases.
5. **Profit Volatility & Outliers:** The Profit boxplot indicates that while most transactions cluster tightly around small positive margins, extreme negative outliers still occur on heavily discounted items.
